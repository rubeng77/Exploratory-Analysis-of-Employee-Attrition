# Confidence Intervals

## Confidence Interval for Mean Monthly Income

```python
from scipy.stats import t

n = len(IBM['MonthlyIncome'])
sample_mean = IBM['MonthlyIncome'].mean()
std_dev = IBM['MonthlyIncome'].std(ddof=1)
confidence = 0.95
alpha = 1 - confidence

df = n - 1

t_critical = t.ppf(1 - (alpha/2), df)

se = std_dev / np.sqrt(n)

E = t_critical * se

ci_lower = sample_mean - E
ci_upper = sample_mean + E

print(f"Sample Mean: {sample_mean:.2f}")
print(f"Sample Standard Deviation: {std_dev:.2f}")
print(f"Degrees of Freedom: {df}")
print(f"Critical t-value: {t_critical:.3f}")
print(f"Margin of Error: {E:.2f}")
print(f"95% Confidence Interval for Mean Monthly Income: [{ci_lower:.2f}, {ci_upper:.2f}]")
```

I used the t-distribution because the population standard deviation is unknown. This dataset does not have the true standard deviation of the entire IBM employee population. The 95% confidence interval given tells us that, based on our sample of 1,470 employees, we can be 95% confident that the true average monthly income for the population of employees is between 6,262.06 - 6,743.80.

## Confidence Interval for Attrition Proportion

```python
from scipy.stats import norm

n = len(IBM)

total_left = IBM['Attrition'] == 1
num_attrition = len(IBM[total_left])

p_hat = num_attrition / n

print(f"Sample Proportion: {p_hat:.3f}")
print(f"Result for np̂: {n * p_hat:.2f}")
print(f"Result for n(1-p̂): {n * (1 - p_hat):.2f}")
```

```python
SE = np.sqrt((p_hat * (1 - p_hat)) / n)

confidence_level = 0.95
ci_lower, ci_upper = norm.interval(confidence_level, loc=p_hat, scale=SE)

print(f"95% Confidence Interval for the Attrition Proportion: ({ci_lower:.3f}, {ci_upper:.3f})")
```

I use the normal distribution here because our sample size is large enough and both np^ = 237 and n(1-p^) = 1,233 are greater than 10. This confirms I can use the normal distribution for this calculation. The 95% confidence interval given tells us that, based on our sample of
1,470 employees, we can be 95% confident that the true attrition proportion for the population of employees is between 14.2% and 18.0%.

# Hypothesis Testing

## One-Sample T-Test: Mean Monthly Income

Lets say we want to see if the average monthly income of employees in this dataset is significantly different from 6,500. I am using 6,500 as it seems like it's somewhat in the middle based on the past analysis. Since the population standard deviation is unknown, we will use the t-distribution. This is a two-tailed test with a significance level of .05.
- Null Hypothesis (H0): The true mean monthly income of employees is $6,500

- Alternative Hypothesis (H1): The true mean monthly income of employees
  is different from $6,500


```python
from scipy.stats import norm, t

xbar = IBM['MonthlyIncome'].mean()
n = len(IBM['MonthlyIncome'])
s = IBM['MonthlyIncome'].std()
mu_0 = 6500

t_statistic = (xbar - mu_0) / (s / np.sqrt(n))
print(f"T-statistic: {t_statistic:.4f}")

degrees_of_freedom = n - 1
p_value = 2 * (1 - t.cdf(abs(t_statistic), df=degrees_of_freedom))
print(f"P-value: {p_value:.4f}")

alpha = 0.05
if p_value < alpha:
    print("Conclusion: Reject the null hypothesis.")
    print("Interpretation: There is sufficient statistical evidence at the 5% significance level to conclude that the mean monthly income is significantly different from $6,500.")
else:
    print("Conclusion: Fail to reject the null hypothesis.")
    print("Interpretation: There is not sufficient statistical evidence at the 5% significance level to conclude that the mean monthly income is significantly different from $6,500.")
```

We fail to reject the null hypothesis. With a p-value of 0.9810, which is greater than our significance level of 0.05. There is not enough
statistical evidence to say that the true mean monthly income is
significantly different from $6,500.

## Two-Sample T-Test: Monthly Income by Attrition

Lets say we want to see if employees who left the company earn significantly different monthly incomes compared to employees who stayed. Since the population standard deviation is unknown, we will use the t-distribution. This is a two-tailed test with a significance level of 0.05.
- Null Hypothesis (H0): The mean monthly income of employees who left is equal to the mean monthly income of employees who stayed
- Alternative Hypothesis (H1): The mean monthly income of employees who left is different from the mean monthly income of employees who stayed

```python
from scipy.stats import ttest_ind

left = IBM[IBM['Attrition'] == 1]['MonthlyIncome']
stayed = IBM[IBM['Attrition'] == 0]['MonthlyIncome']

t_statistic, p_value = ttest_ind(left, stayed, equal_var=False)

print(f"T-statistic: {t_statistic:.4f}")
print(f"P-value: {p_value:.4f}")

alpha = 0.05
if p_value < alpha:
    print("Conclusion: Reject the null hypothesis.")
    print("Interpretation: There is sufficient statistical evidence at the 5% significance level to conclude that employees who left earn significantly different incomes than those who stayed.")
else:
    print("Conclusion: Fail to reject the null hypothesis.")
    print("Interpretation: There is not sufficient statistical evidence at the 5% significance level to conclude that employees who left earn significantly different incomes than those who stayed.")
```

We reject the null hypothesis. With a p-value of 0.0000, which is less
than our significance level of 0.05. There is enough statistical evidence to conclude that employees who left the company earn significantly different monthly incomes than employees who stayed.

## One-Sample Proportion Test: Attrition Rate

Lets say we want to see if the attrition rate in this dataset is
significantly different from 15%. I am using 15% as its the average attrition rate when you look it up on Google. This is a
two-tailed test with a significance level of 0.05.

- Null Hypothesis (H0): The true attrition rate of employees is 15%
- Alternative Hypothesis (H1): The true attrition rate of employees
  is different from 15%

```python
from statsmodels.stats.proportion import proportions_ztest

total_left = IBM['Attrition'] == 1
count = len(IBM[total_left])
nobs = len(IBM)

stat, pval = proportions_ztest(count, nobs, value=0.15, alternative='two-sided')

print(f"Z-stat: {stat:.4f}, p-value: {pval:.4f}")
```

We fail to reject the null hypothesis. With a p-value of 0.2419, which is greater than our significance level of 0.05. There is not enough statistical evidence to conclude that the attrition rate is significantly different from 15%. This suggests that the attrition rate in this dataset could be 15%.

# Chi-square Tests

##Chi-Square Test of Independence of Departmnet and Attrition

Lets say we want to explore whether there is a significant association between the department an employee works at and whether they left. Sales appeared to have a higher attrition rate baased on our earlier bar chart. This test will determine if the relationship between Department and Attrition is statistically significant or just random chance.
This is a two-tailed test with a significance level of 0.05.

- Null Hypothesis (H0): Department and Attrition are independent
- Alternative Hypothesis (H1): Department and Attrition are not
  independent

```python
from scipy.stats import chi2_contingency

contingency_table = pd.crosstab(IBM['Department'], IBM['Attrition'])
contingency_table
```

```python
chi2, p, dof, expected = chi2_contingency(contingency_table)

print("Chi-square Statistic:", chi2)
print("Degrees of Freedom:", dof)
print("P-value:", p)

print("\nObserved values:")
print(contingency_table)

print("\nExpected values:")
print(pd.DataFrame(expected, index=contingency_table.index, columns=contingency_table.columns))

alpha = 0.05
if p < alpha:
    print("\nConclusion: Reject the null hypothesis.")

else:
    print("\nConclusion: Fail to reject the null hypothesis.")
```

We reject the null hypothesis. With a p-value of 0.0045, which is less
than our significance level of 0.05. There is enough statistical evidence to say that there is a significant association between department and attrition. This means that the department an employee works in does play a role in whether they leave.

## Chi-Square Test of Homogeneity of Job Satisfaction Distribution Across Departmet

Lets say we want to explore whether the distribution of job satisfaction
levels is the same across all three departments, which are Human Resources, Research & Development, and Sales. If job satisfaction is distributed differently across departments, it would help to explain why some departments have higher attrition rates than others.
This is a two-tailed test with a significance level of 0.05.

- Null Hypothesis (H0): The distribution of job satisfaction levels
  is the same across all departments
- Alternative Hypothesis (H1): The distribution of job satisfaction
  levels is different across at least one department



```python
contingency_table2 = pd.crosstab(IBM['Department'], IBM['JobSatisfaction'])

chi2_2, p_2, dof_2, expected_2 = chi2_contingency(contingency_table2)

print("Chi-square Statistic:", chi2_2)
print("Degrees of Freedom:", dof_2)
print("P-value:", p_2)

print("\nObserved values:")
print(contingency_table2)

print("\nExpected values:")
print(pd.DataFrame(expected_2, index=contingency_table2.index, columns=contingency_table2.columns))

alpha = 0.05
if p_2 < alpha:
    print("\nConclusion: Reject the null hypothesis.")

else:
    print("\nConclusion: Fail to reject the null hypothesis.")

```

We fail to reject the null hypothesis. With a p-value of 0.2019, which is greater than our significance level of 0.05. There is not enough statistical evidence to say that job satisfaction levels are distributed differently across departments. This suggests that employees across departments usually report similar levels of job satisfaction regardless of which department they work in.
